```python
import pandas as pd
import sys
import os

def inspeccionar_parquet(ruta_archivo):
    """
    Carga un archivo Parquet y muestra un resumen de su contenido.
    """
    # 1. Validar existencia del archivo
    if not os.path.exists(ruta_archivo):
        print(f"❌ Error: El archivo '{ruta_archivo}' no existe.")
        return

    try:
        # 2. Cargar el archivo usando el motor pyarrow (por defecto en pandas recientes)
        print(f"📂 Cargando: {ruta_archivo}...")
        df = pd.read_parquet(ruta_archivo)

        # 3. Imprimir Información General
        print("-" * 50)
        print("RESUMEN DEL ARCHIVO")
        print("-" * 50)
        print(f"Dimensiones: {df.shape[0]} filas x {df.shape[1]} columnas")
        
        print("\n--- Tipos de Datos (Dtypes) ---")
        print(df.dtypes)
        
        # 4. Mostrar una muestra de datos
        print("\n--- Vista Previa (Primeras 5 filas) ---")
        print(df.head())
        
        # Opcional: Mostrar estadísticas básicas si hay datos numéricos
        print("\n--- Estadísticas Básicas ---")
        print(df.describe())

    except Exception as e:
        print(f"❌ Ocurrió un error al leer el archivo: {e}")

```
